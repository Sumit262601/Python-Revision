```python
#  Tuple Packing
# Tuple packing in Python is the process of assigning multiple comma-separated values to a single variable, creating a tuple.

student = "Sumit", True, 34, 33.4
print(student)
```

```python
# Unpacking
# Tuple unpacking allows you to extract individual values from a tuple and assign them directly to separate variables

name, general, age, marks = student
print(name)
print(general)
print(age)
print(marks)
```

```python
# Find the second occurrence index of 20 without converting the tuple into another data structure.

t = (10, 20, 10, 30, 40, 20, 10, 50)

# first occur of 20 is index 1 and second occur of 20 is index 5
first_index = t.index(20)

second_index = t.index(20, first_index + 1)
print(second_index)
```

```python
students = (

    ("A", 70),

    ("B", 85),

    ("C", 60),

    ("D", 95),

    ("E", 80)
)

total = 0

for i in range(len(students)):
    total += students[i][1]
    average = total / len(students)
    

print(f"Average of all students is = {average}")

for i in range(len(students)):
    if students[i][1] > average:
        print( students[i][0], students[i][1])
```

```python
num = (
    (10, 20, 30),
    (40, 50, 60),
    (70, 80, 90),
)

for i in range(len(num)):
    sum_num = 0
    for j in range(len(num)):
        sum_num += num[i][j]
    print(sum_num, end=" ")
    print()
```

```python
num = (
    (10, 20, 30),
    (40, 50, 60),
    (70, 80, 90)
)
for i in range(len(num)):
    row_max = num[i][0]

    for j in range(len(num)):
        if num[i][j] > row_max:
            row_max = num[i][j]
    print(row_max)
```
