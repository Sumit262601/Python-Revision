```python
# RIGHT SIDE RIGHT ANGLE TRIANGLE
num = int(input("Enter number of rows: "))

for i in range(1, num):
    rows = 1
    for j in range(num, 0, -1):
        if j > i:
            print(" ", end=" ")
        else:
            print(rows, end=" ")
            rows += 1
    print()

# print()

for i in range(num, 0, -1):
    rows = 1
    for j in range(num, 0, -1):
        if j > i:
            print(" ", end=" ")
        else:
            print(rows, end=" ")
            rows += 1
    print()
```
