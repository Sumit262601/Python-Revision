```python
# What is Broadcasting?

"""
Broadcasting in NumPy allows us to perform arithmetic operations on arrays of different 
shapes without reshaping them. It automatically adjusts the smaller array to match the larger 
array's shape by replicating its values along the necessary dimensions. This makes element-wise 
operations more efficient by reducing memory usage and eliminating the need for loops.
"""

import numpy as np
```

```python
import numpy as np

# Original Array
arr = np.array([10, 20, 30])

# Create a view and a copy
arr_view = arr.view()
arr_copy = arr.copy()

# Modify the original array
arr[0] = 99

print(arr_view) # Output: [99, 20, 30] (Changes reflected)
print(arr_copy) # Output: [10, 20, 30] (Stays independent)

# Check data ownership
print(arr_view.base) # Output: [99, 20, 30] (Points to parent array)
print(arr_copy.base) # Output: None
```
