1. A company wants to generate data for 72 employees. 
• Employee IDs should start from 1001. 
• Employee performance scores should range equally from 35 to 95. 
• Arrange the data into 6 departments with 12 employees each. 
• Extract the last 3 employees from the first 3 departments. 
• Create a separate independent array from the extracted data. 
• Change one value in the new array to 999. 
• Verify whether the original data changed.

```python
emp_id = np.arange(1001, 1001+72)
print("Emplyees Id =", emp_id, "\n")

perf_scores = np.linspace(35, 95, 72)
print("Performance scores =", perf_scores, "\n")

depart_id= emp_id.reshape(6, 12)
depart_scores = perf_scores.reshape(6, 12)
print("Department ID = ", depart_id, "\n")
print("Department scores = ", depart_scores, "\n")

print("Original Performance Scores Shape:", depart_scores.shape, "\n")

extrac_data = depart_scores[:3, 9:]
print("Extracted Data", extrac_data, "\n")

indep_array = extrac_data.copy()
print(indep_array, "\n")

indep_array[0, 0] = 999
print("Modify Independent Array = ", indep_array, "\n")

print("Original Data", depart_scores[0, 9], "\n")
print("Changes Data", depart_scores[0, 9] == 999)
```
