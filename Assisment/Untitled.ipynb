```python
# 1. What is a list in Python? How is it different from a tuple?

A list in Python is a build-in, ordered and mutable data structure used to store a collection of items in a single
variable. Lists are defined using '[]' and can hold elements of different data type (int, str, float, or, even
other lists), including duplicate values.

KEY POINTS OF A LISTS: 
1. Ordered: Items maintain a defin sequence and are accessed via zero-based indexing (list[0]).
2. Mutable: You can add, remove, or modif elements in place after the list is created without creating a new objects in memory.
3. Hetrogeneous & Dynamic: Cna store mixed data types and automatically grows or shrinks as items are added or removed.

Difference Between a List and a Tuple
While bothe lists and tuples are ordered sequences that can store mixed data types and duplicate elements, their
primary difference in mutability: List can be modified whereas tuple cannot.

FEATURE LIST AND TUPLE:
List -> Mutable, Define '[1, 2, 3]', Consumes more memory (allocated extra space for dynamic sizing), Many methods (append(), remvoe(), pop(), sort(), insert(), count(), etc.)
Tuple -> Immutable Define '(1, 2, 3)', Consumes less memory (fixed size is allocated at creation), Only two main methods (count() and index()), 
```

```python
# 20. Write a program that shows the difference between .append() and .extend() when adding a list [7, 8, 9] to [1, 2, 3].

# Using .append()
list_append = [1, 2, 3]
list_append.append([7, 8, 9])
print("Using .append():", list_append)
print("Length after .append():", len(list_append))

print("-" * 40)

# Using .extend()
list_extend = [1, 2, 3]
list_extend.extend([7, 8, 9])
print("Using .extend():", list_extend)
print("Length after .extend():", len(list_extend))
```

```python
# 22. Write a program to add a new element to a list only if it doesn't already exist in the list.

nums = [10, 20, 30, 40, 50]
new_element = 60

if new_element not in nums:
    nums.append(new_element)
    print(f"{new_element} added to the list.")
else:
    print(f"{new_element} already exists in the list.")

print("Updated list:", nums)
```

```python
# 23. Take 5 numbers as input from the user and add each one to a list using .append() inside a loop.

numbers = []

for i in range(5):
    num = int(input(f"Enter number {i + 1}: "))
    numbers.append(num)

print("List of numbers:", numbers)
```
