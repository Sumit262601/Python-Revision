```python
num = int(input("Enter the number: "))
largest = 0
smallest = 0

for d in range(9, 0, -1):
    t = num
    while t > 0:
        if t % 10 == d:
            largest = largest * 10 + d
        t //= 10

for d in range(1, 10):
    t = num 
    while t > 0:
        if t % 10 == d:
            smallest = smallest * 10 + d
        t //= 10

print("Largest: ", largest)
print("Smallest: ", smallest)
print("Difference", largest - smallest)
```

```python
num = int(input("Enter the number: "))

temp = num
total_divisor = 1
fact_str = ""
divisor = 2
first_fact = True

while temp > 1:
    count = 0

    while temp % divisor == 0:
        count += 1
        temp //= divisor

    if count > 0:
        total_divisor *= count + 1
    
        if not first_fact:
            fact_str += " x "
        fact_str += str(divisor) + '^' + str(count)
        first_fact = False

    divisor += 1

print(f"Output: {fact_str}")
print(f"Total divisor: {total_divisor}")
```

```python
start = int(input("Start: "))
end = int(input("End: "))

count = 0

for p in range(max(start, 2), end - 1):
    q = p + 2

    p_prime = True
    i = 2
    while i * i <= p:
        if p % i == 0:
            p_prime = False
            break
        i += 1

    q_prime = True
    i = 2
    while i * i <= q:
        if q % i  == 0:
            q_prime = False
            break
        i += 1

    if p_prime and q_prime:
        print(f"({p}, {q})", end = " ")
        count += 1
    
print("Total paris = ", count)
```
