```python
a = [1, 2, 3, 4, 5, 6, 7, 8, 9, 18, 17]
b = [9, 8, 7, 6, 10, 12, 14, 15, 17]
c = [1, 2, 3, 4, 5, 6, 16, 19, 17, 20]
common_num = set()

for i in range(len(a)):
    for j in range(len(b)):
        for k in range(len(c)):
            if a[i] == b[j] == c[k]:
                common_num.add(a[i])
            
print(common_num)
```

```python
a = [1, 2, 3, 4, 5, 6, 7, 8, 9, 18, 17]
b = [9, 8, 7, 6, 10, 12, 14, 15, 17]
c = [1, 2, 3, 4, 5, 6, 16, 19, 17, 20]

combined_lists = a + b + c
unique_num = set()

for num in combined_lists:
    if combined_lists.count(num) == 1:
        unique_num.add(num)

print(unique_num)
```

```python
number = [1, 2, 3, 4, 5, 6, 7, 8, 9, 18, 17, 3, 4, 2, 6, 7, 10]
frequency = set()
count = 0

for num in set(number):
    if number.count(num) > 1:
        frequency.add(num)
        count += 1

print(f"Frequency of the number in the list: {frequency} : {count}")
```

```python
lst = [1, 2, 3, 2, 4, 2, 5, 3, 3]
max_count = 0
result = set()

for i in lst:
    count = 0

    for j in lst:
        if i == j:
            count += 1

    if count > max_count:
        max_count = count
        result = {i}

    elif count == max_count:
        result.add(i)

print("Maximum frequency:", max_count)
print("Numbers:", result)
```

```python
# UNION
Name = {"Atul", "Sumit", "Rahul", "Anjali"}
Name2 = {"Sumit", "Rohit", "Roshni", "Atul"}

result =  Name.union(Name2) #Name | Name2
print(result)
```

```python
# INTERSECTION
Name = {"Atul", "Sumit", "Rahul", "Anjali"}
Name2 = {"Sumit", "Rohit", "Roshni", "Atul"}

result =  Name.intersection(Name2) #Name & Name2
print(result)
```

```python
python = {"Aman", "Riya", "Rahul", "Neha"}
sql = {"Rahul", "Neha", "Vikas", "Priya"}
powerbi = {"Neha", "Vikas", "Priya", "Karan"}

# Find all unique students enrolled in at least one course.
result = python | sql | powerbi
print(result)
```

```python
# Find students who are enrolled in both Python and SQL.
result = python | sql
print(result)
```

```python
# Find students who are enrolled in SQL and Power BI.
result = sql | powerbi
print(result)
```

```python
# Find students who are enrolled in all three courses.

result = sql & powerbi & python
print(result)
```

```python
# Find students who are enrolled in Python but not SQL or Power BI.

result = python - (sql.union(powerbi))
print(result)
```

```python
# Find students who are enrolled in SQL but not Python.

result = sql.intersection(python)
print(result)
```

```python
# Find students who are enrolled in Power BI but not Python.

result = powerbi.intersection(python)
print(result)
```
