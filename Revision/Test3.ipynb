Given: L = [12, 25, 12, 40, 55, 25, 70, 12, 90]

Perform the following:

Find all unique values.
Create a duplicate-free list while maintaining the original order.
Sort the final list in descending order.
Find the frequency of each value using a dictionary.

```python
L = [12, 25, 12, 40, 55, 25, 70, 12, 90]

unique_values = set(L)
print("Unique values:", unique_values)

unique_ordered_list = list(dict.fromkeys(L))
print("Duplicate-free list (ordered):", unique_ordered_list)

unique_ordered_list.sort(reverse=True)
print("Sorted list (descending):", unique_ordered_list)

frequency_dict = {}
for item in L:
    frequency_dict[item] = frequency_dict.get(item, 0) + 1

print("Frequency of each value:", frequency_dict)
```

5. Generate 60 equally spaced values between 10 and 100 using np.linspace().

Reshape them into:

5 × 4 × 3

Then extract:

Last 2 layers
First 3 rows
Alternate columns

Create a copy of the result and change one element to 999.

Check whether the original array changes.

6.

Given:

T = (10, 20, 30, 40, 50, 60, 20, 30)

Perform:

Find length
Find maximum and minimum
Count 20
Find index of 40
Extract alternate elements
Extract last 4 elements
Convert to list
Add 100
Convert back to tuple

```python
T = (10, 20, 30, 40, 50, 60, 20, 30)

print("Length of the Tuple items: ", len(T))
print("Maximum number of the Tuple: ", max(T))
print("Count of the 20 in the Tuple: ", T.count(20))
print("Index of 40 items in Tuple: ", T.index(40))
print("Extract alternate element of the tuple: ", T[::2])
print("Extract the lst 4 element: ", T[-4:])
print("Convert the tuple into list: ")
New_T = list(T)
print("Convert the tuple into list: ", New_T)
New_T.append(100)
print(New_T)

print("Convert the tuple into list: ")
T = tuple(New_T)
print(T)
```

```python
L = [10,20,30,40,50]
arr = np.array([10, 20, 30, 40, 50])

New_L = L[1:4]
print(New_L, "\n")

New_L[0] = 100
print(New_L, "\n")

new_arr = arr[1:4]
print("Original slice:", new_arr)

new_arr[0] = 100

print("Modified slice:", new_arr)
print("Original array (arr):", arr)
```
