```python
# Q1. Number Frequency Analyzer Take N integers from the user and store them in a list. Print: • 

# The numbers that occur exactly once.

lst = eval(input("Enter the list: "))
count = 0
uniq_num = []
occur = []
freq = []


for num in lst:
    if lst.count(num) == 1:
        uniq_num.append(num)
    
print(f"Unique number: {uniq_num}")


# The numbers that occur more than once.

for num in lst:
    if lst.count(num) > 1 and num not in occur:
        occur.append(num)

print(f"Occur number is: {occur}")


# The number having the highest frequency.

for num in lst:
    if lst.count(num) > count:
        count = lst.count(num)
        freq = num

print(f"Frequenty of the list: {freq} and {count}")
```

```python
# Q3. Student Performance Dictionary Create a dictionary containing student names and their marks in 3 subjects:

students = { 
    "Rahul": (78, 85, 91), 
    "Aman": (45, 67, 52), 
    "Priya": (92, 88, 95), 
    "Neha": (55, 49, 61)
}
total_marks = {}


# Total marks of every student 
avg_marks = {}
for name, marks in students.items():
    curr_total = 0
    for mark in marks:
        curr_total += mark

    total_marks[name] = curr_total
    
print('======= Total marks of every student  ======')
for name, total in total_marks.items():
    print(f"Students of {name} and marks {total}")
    
print()


# Average marks of every student
for name, marks in students.items():
    curr_total = 0
    sub_count = 0
    
    for mark in marks:
        curr_total += mark
        sub_count += 1

    average = curr_total / sub_count
    avg_marks[name] = average
    
print('======= Average marks of every student ======')
for name, avg in avg_marks.items():
   
    print(f"Average of {name} their {avg} ")

print()

# Student with the highest total
highest = 0
top_student = None

for name, marks in students.items():
    current_total = 0
    
    for mark in marks:
        current_total += mark
        
    if current_total > highest:
        highest = current_total
        top_student = name
print('======= Student with the highest total ======')
print(f"Students {top_student} with their highest markrs {highest}")

print()


# Students who passed all subjects
passing_mark = 50
passed_students = []

print('======= Students who passed all subjects ======')
for name, marks in students.items():
    passed_all = True
    
    for mark in marks:
        if mark < passing_mark:
            passed_all = False
            break

    if passed_all:
        passed_students.append(name)

print(passed_students)
```

```python
# Q4. Common & Unique Elements Take two lists from the user.

lst1 = [10, 20, 30, 40, 50]
lst2 = [30, 40, 60, 70, 80]

# 1. Elements present in both lists 
lst3 = []
for i in range(len(lst1)):
    for j in range(len(lst2)):
        if lst1[i] == lst2[j]:
            lst3.append(lst1[i])
            
print(lst3)

# 2. Elements present only in the first list 
c_lst1 = []
for item in lst1:
    if item not in lst2:
        c_lst1.append(item)

print(c_lst1)

# 3. Elements present only in the second list 
c_lst2 = []
for item in lst2:
    if item not in lst1:
        c_lst2.append(item)
        
print(c_lst2)

# 4. Elements present in either list but not both Do not print duplicate values in the output.
lst4 = []

lst4 = c_lst1 + c_lst2
print(lst4)
```

```python
# Q5. Compress Consecutive Characters Take a string from the user and count consecutive repeated characters.
# Example: Input: aaabbccccdaa Output: a3b2c4d1a2

string = input("Enter the string: ")
count = 1
compres_string = ""

if len(string) > 0:
    for i in range(1, len(string)):
        if string[i] == string[i - 1]:
            count += 1
        else:
            compres_string += string[i-1] + str(count)
            count = 1
            
    compres_string += string[-1] + str(count)
    
print(f"Output: {compres_string}")
```

```python
# Q7. Pattern Challenge Write a Python program to print the following pattern for n = 5:

num = int(input("Enter the number: "))

for i in range(num):
    for j in range(i + 1):
        print("* ", end = " ")
    print("")

for i in range(num):
    for j in range(i, num - 1):
        print("* ", end = " ")
    print("")

        
```

```python
# Take a positive integer from the user. 
# Determine whether the number is a special number. 
# A number is special if: Sum of its digits + Product of its digits = Original number

num = int(input("Enter the number: "))

temp = num
digit_sum = 0
digit_product = 1

while temp > 0:
    digit = temp % 10
    digit_sum += digit
    digit_product *= digit  
    temp //= 10

total = digit_sum + digit_product


if total == num:
    print(f"For {num} → sum = {digit_sum}, product = {digit_product}, so {digit_sum} + {digit_product} = {total}. Therefore, it is Special.")
else:
    print(f"For {num} → sum = {digit_sum}, product = {digit_product}, so {digit_sum} + {digit_product} = {total}. Therefore, it is Not Special.")
```

```python
# Q8. Dictionary InversionGiven: 
data = { 
    "A": "Python", 
    "B": "Java", 
    "C": "Python", 
    "D": "C++", 
    "E": "Java" 
}

pair_data = {}

for key, value in data.items():
    if value not in pair_data:
        pair_data[value] = [key]
    else:
        pair_data[value].append(key)

print(pair_data)
```

```python
# Q10. Shopping Cart Analyzer — Challenge Create a shopping cart using a dictionary: 
cart = { 
    "Laptop": (2, 50000), 
    "Mouse": (3, 800), 
    "Keyboard": (1, 1500), 
    "Monitor": (2, 12000) 
}

total_bill = 0
highest_cost = 0
most_expensive_product = ""
products_above_10k = []

print("--- 1. Total cost of each product ---")
for product, (quantity, price) in cart.items():
    total_cost = quantity * price
    print(f"{product}: {total_cost}")
    
    total_bill += total_cost
    
    if total_cost > highest_cost:
        highest_cost = total_cost
        most_expensive_product = product
        
    if total_cost > 10000:
        products_above_10k.append(product)

print("\n--- 4. Products > 10,000 ---")
print(products_above_10k)

if total_bill >= 100000:
    discount_percent = 15
elif total_bill >= 50000:
    discount_percent = 10
else:
    discount_percent = 0

# Calculate final amounts
discount_amount = total_bill * (discount_percent / 100)
final_bill = total_bill - discount_amount

# Final Output
print("\n--- Final Summary ---")
print(f"Total Bill: {total_bill}")
print(f"Discount Applied: {discount_amount} ({discount_percent}%)")
print(f"Final Bill: {final_bill}")
print(f"Most Expensive Product: {most_expensive_product}")
```

```python
# Q2. Hollow Diamond Pattern
# Take an odd integer n from the user and print a hollow diamond using *.

n = int(input("Enter an odd integer n for the total height: "))

if n % 2 == 0:
    print("Please enter an odd number.")
else:
    rows = (n // 2) + 1

    ## HOLLOW UPPER TRIANGLE
    for i in range(1, rows + 1): 
        # Print leading spaces
        for space in range(1, (rows - i) + 1): 
            print("  ", end="")
            
        # Print stars only at the borders, spaces inside
        for j in range(1, 2 * i):
            if j == 1 or j == (2 * i - 1):
                print("* ", end="")
            else:
                print("  ", end="")
        print()

    ## HOLLOW INVERTED TRIANGLE
    for i in range(rows - 1, 0, -1):
        # Print leading spaces
        for space in range(1, (rows - i) + 1):
            print("  ", end="")
            
        # Print stars only at the borders, spaces inside
        for j in range(1, 2 * i):
            if j == 1 or j == (2 * i - 1):
                print("* ", end="")
            else:
                print("  ", end="")
        print()
```
