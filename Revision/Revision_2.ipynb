```python
# 1. Take a string and find the first character that appears only once. If no such character exists, print "No unique character".
char = input("Enter the string: ")
chars = []
counts = []

for ch in char:
    if ch in chars:
        idx = chars.index(ch)
        counts[idx] += 1
    else:
        chars.append(ch)
        counts.append(1)

result = None
for i in range(len(chars)):
    if counts[i] == 1:
        result = chars[i]
        break

if result is not None:
    print("First character:", result)
else:
    print("No unique character")
```

```python
# 3. Take a number and determine whether it is an Armstrong number.

num = int(input("Enter the number: "))
temp = num
sum_var = 0

power_var = len(str(num))

while temp > 0:
    digit = temp % 1014
    sum_var += digit ** power_var
    temp //= 10

if(sum_var == num):
    print(f"This is an armstorm number: {num}")
else:
    print(f"This is not an armstorm number: {num}")
```

```python
# 4. Given [10, 20, 10, 30, 20, 40, 10], print the frequency of every element without using count().

lst = [10, 20, 10, 30, 20, 40, 10]
unique_num = []
frequency = []

for num in lst:
    if num in unique_num:
        index = unique_num.index(num)
        frequency[index] += 1

    else:
        unique_num.append(num)
        frequency.append(1)

for i in range(len(unique_num)):
    print(f"Element {unique_num[i]} = frequency {frequency[i]}")
```

```python
# 8. Given [0, 5, 0, 3, 8, 0, 2], move all 0s to the end while maintaining the order of the remaining elements. Expected: [5, 3, 8, 2, 0, 0, 0].

lst = [0, 5, 0, 3, 8, 0, 2]
length = len(lst)
end = length - 1
i = 0

while i <= end:
    ele = lst [i]
    if ele == 0:
        for j in range(i, end):
            lst[j] = lst[j + 1]

        lst[end] = 0
        end -= 1

    else:
        i += 1

print(f"All zerors in right side {lst}")
```

```python
#9. Take a string and check whether it is a palindrome, ignoring spaces and case. Example: "Never Odd Or Even".

text = input("Enter the string value: ")

# Space removing and lowercasing
cleaned = ""
for ch in text:
    if ch != " ":
        cleaned += ch.lower()

# Palindrome check
is_palindrome = True
left = 0
right = len(cleaned) - 1

while left < right:
    if cleaned[left] != cleaned[right]:
        is_palindrome = False
        break
    left += 1
    right -= 1
    
if is_palindrome:
    print(f"This string is a palindrome: {text}")
    print(f"Without spacing or case {cleaned}")
else: 
    print(f"This is not a palindrome: {text}")
```

```python
# 20. Find the missing number from a list containing numbers from 1 to n. Example: [1,2,3,5,6] → 4.

lst = [1, 2, 3, 5, 6]
missing = []

for i in range(1, len(lst)):
    diff = lst[i] - lst[i -1]

    if diff > 1:
        for j in range(1, diff):
            missing.append(lst[i-1] + j)

print(f"Missing Number is: {missing}")
```

```python
# 22. Take a number and check whether it is a strong number. Example: 145 → 1! + 4! + 5! = 145.

num = int(input("Enter the number: "))
n = num
sum_fact = 0

while n > 0:
    digit = n % 10
    n //= 10

    i = 1
    fact = 1
    while i <= digit:
        fact *= i
        i += 1

    sum_fact += fact
    print(f"Factorial of number: {digit} - {fact}")

if sum_fact == num:
    print(f"\nSum of all factorial number: {sum_fact}")
else:
    print(f"\nThis ia not a sum of factorial number: {sum_fact}") 
```

```python
# 34. Given a list, find the length of the longest consecutive sequence. Example: [100,4,200,1,3,2] → 4, because 1,2,3,4 is a sequence.

nums = [100, 4, 200, 1, 3, 2]
num_set = set(nums)
longest_streak = 0

for num in num_set:
    if num - 1 not in num_set:
        current_num = num
        current_streak = 1

        while current_num + 1 in num_set:
            current_num += 1
            current_streak += 1

        if current_streak > longest_streak:
            longest_streak = current_streak

print(f"Longest sequence length: {longest_streak}")
```

```python
# 35. Take a string and count vowels, consonants, digits and spaces.

text = "Python is Fun! Let's code 123"

vowels_count = 0
consonants_count = 0
digits_count = 0
spaces_count = 0

vowel_chars = "aeiouAEIOU"

for char in text:
    if char in vowel_chars:
        vowels_count += 1
    elif char.isalpha():
        consonants_count += 1
    elif char.isdigit():
        digits_count += 1
    elif char.isspace():
        spaces_count += 1

print(f"String: '{text}'")
print(f"Vowels: {vowels_count}")
print(f"Consonants: {consonants_count}")
print(f"Digits: {digits_count}")
print(f"Spaces: {spaces_count}")
```

```python
# 36. Given numbers = [2,4,3,5,7,8] and target = 10, find all pairs whose sum is equal to the target. Expected pairs include 2,8 and 3,7. 

numbers = [2, 4, 3, 5, 7, 8]
target = 10

pairs = []
seen = set()

for num in numbers:
    complement = target - num
    
    if complement in seen:
        pairs.append((complement, num))
        
    seen.add(num)

print(f"Pairs with sum {target}: {pairs}")
```

```python
# 38. Take a string and find the most frequent character without using count(). 

text = "success"
char_frequency = {}
most_frequent_char = ""
max_count = 0

for char in text:
    if char in char_frequency:
        char_frequency[char] += 1
    else:
        char_frequency[char] = 1

for char, count in char_frequency.items():
    if count > max_count:
        max_count = count
        most_frequent_char = char

print(f"String: '{text}'")
print(f"Most frequent character: '{most_frequent_char}' frequent {max_count} times")
```
