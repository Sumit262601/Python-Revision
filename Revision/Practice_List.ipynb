```python
#Program to find the frequencies of all elements of a list. Also, print the list of unique elements in the list and duplicate elements in the given list.

lst = eval(input("Enter list: "))
length = len(lst)
uniq = []
dupl = []
count = i = 0

while i < length:
    element = lst[i]
    count = 1

    if element not in uniq and element not in dupl:
        i += 1

        for j in range(i, length):
            if element == lst[j]:
                count += 1
        
        else: 
            print(f"Element {element} with frequiency {count}")

            if count == 1:
                uniq.append(element)
            else:
                dupl.append(element)

    else:
        i += 1

print(f"Original list: {lst}")
print(f"Unique list: {uniq}")
print(f"Duplicate list: {dupl}")
```

```python
# Find largest consecutive number

a = [2, 100, 1, 0, 3, 4, 5]
a.sort()

longest_seq = []
current_seq = []

for i in range(len(a)):
    if a[i] == a[i - 1] + 1:
        current_seq.append(a[i])

    elif a[i] != a[i-1]:
        if len(current_seq) > len(longest_seq):
            longest_seq = current_seq
        current_seq = [a[i]]

if len(current_seq) > len(longest_seq):
    longest_seq = current_seq

print(f"largest consecutive number: {longest_seq}")
print(f"Length: {len(longest_seq)}")
```

```python
lst = eval(input("Enter the list: "))
lst = list(lst) 
length = len(lst)
missing_num = []


for i in range(1, length):
    diff = lst[i] - lst[i - 1]
    if diff > 1:
        for j in range(1, diff):
            missing_num.append(lst[i - 1] + j)

print(f"Missing numbers are: {missing_num}")
```
