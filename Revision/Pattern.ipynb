```python
# PERFECT TRIANGLE
rows = int(input("Enter number of rows: "))
k = 0

for i in range(1, rows + 1):
    for space in range(1, (rows-i)+1): # for spacing the
        print(end="  ")

    while k != (2*i-1):
        print("* ", end="")
        k += 1
    k=0
    print()
```

```python
# RIGHT SIDE RIGHT ANGLE TRIANGLE
num = int(input("Enter number of rows: "))

for i in range(1, num):
    rows = 1
    for j in range(num, 0, -1):
        if j > i:
            print(" ", end=" ")
        else:
            print(rows, end=" ")
            rows += 1
    print()
```
