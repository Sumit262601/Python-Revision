```python
num = [10, 20, 30]

print("------------- APPEND ----------------")
num.append(40)
print(num)

print("\n------------- EXTEND ----------------")
num.extend([50, 60])
print(num)

print("\n------------- INSERT ----------------")
num.insert(2, 25)
print(num)

print("\n------------- remove ----------------")
print(num.remove(10))
print(num)

print("\n------------- POP ----------------")
print(num.pop())

print("\n------------- INDEX ----------------")
print(num.index(30))

print("\n------------- COUNT ----------------")
print(num.count(25))

print("\n------------- REVERSE ----------------")
num.reverse()
print(num)

print("\n------------- CLEAR ----------------")
num.clear()
print(num)
```

```python
numbers = list(map(int, input("Enter numbers separated by space: ").split()))

total_sum = sum(numbers)
max_val = max(numbers)
min_val = min(numbers)
avg_val = total_sum / len(numbers)
range_val = max_val - min_val

second_largest = sorted(list(set(numbers)))[-2]

print(f"Sum = {total_sum}, Max = {max_val}, Min = {min_val}")
print(f"Average = {avg_val}, Range = {range_val}")
print(f"Second largest = {second_largest}")
```

```python
words = ["banana", "kiwi", "apple", "fig", "cherry"]

asc_sorted = sorted(words)
desc_sorted = sorted(words, reverse = True)

print("Asceding: ", " ".join(asc_sorted))
print("Descending: ", " ".join(desc_sorted))

len_sorted = sorted(words, key = len)
print("By length: ", " ".join(len_sorted))

words.sort()
words.reverse()
print("Final list:", words)

# `sorted()` returns a **new sorted list** without modifying the original iterable, whereas `.sort()` sorts an existing list **in place** and returns `None`.
```

```python
x = [4, 2, 7, 4, 9, 2, 7, 1, 9, 5]
unique = []
removed = []

print("--------------- UNIQUE LIST ------------------")
for num in x:
    if num not in unique:
        unique.append(num)
    else:
        if num not in removed:
            removed.append(num)

print("Unique list     :", unique)
print("Removed items   :", removed)

print("\n--------------- Repeated frequencies ------------------")
for num in unique:
    freq = x.count(num)
    if freq > 1:
        print(f"{num} -> {freq} times")
```

```python
names = ["Rahul", "Aman", "Priya", "Neha"]
marks = [78, 45, 92, 55]

student_pair = list(zip(marks, names))
ranked_student = sorted(student_pair, reverse = True)

print("Rank List: ")
for i, (m, name) in enumerate(ranked_student, start = 1):
    print(f"Rank {1}: {name} ({m})")

total_marks = sum(marks)
class_avg = total_marks / len(marks)

topper_name = ranked_student[0][1]

above_avg_name = [name for m, name in student_pair if m > class_avg]

print(f"Topper: {topper_name}, Average = {class_avg}")
print("Above average:", ", ".join(above_avg_name))
```
