```python
# Q 2 Number Reversal + Palindrome

num = int(input("Enter the number: "))

n = num
revers_num = 0

while n > 0:
    digit = n % 10
    revers_num = revers_num * 10 + digit
    n //= 10

print(revers_num)

if num == revers_num:
    print("Palidrom number")
else:
    print("Not Palidrom number")
    
```

```python
# Q 4 Strong Number
num = int(input("Enter the number: "))

n = num
sum_fact = 0
while n > 0:
    digit = n % 10
    n //= 10

    i = 1
    fact = 1
    while i <= digit:
        fact *= i
        i += 1

    sum_fact += fact

    print(f"Factorial of number: {digit} - {fact}")

if sum_fact == num:
    print(f"\nSum of all factorial number: {sum_fact}")
else:
    print(f"\nThis ia not a sum of factorial number: {sum_fact}") 
```

```python
# ATM MACHINE

balance = 10000

while True:
    print("\n---------- ATM MENU ----------")
    print("1 Check Balance")
    print("2 Deposit Amount")
    print("3 Withdraw Amount")
    print("4 Exit")

    choice = int(input("Enter the amount: "))

    if choice == 1:
        print(f"Check Balance {balance}")

    elif choice == 2:
        amount = float(input("Enter amount to deposit: "))
        
        if amount > 0:
            balance += amount
            print(f"Deposit amount {amount}")
            print(f"New balance {balance}")
        else:
            print("Invalid deposit amount")

    elif choice == 3:
        amount = float(input("Enter amount to withdraw: "))
        if amount <= 0:
            print("Invalid withdraw amount")
        elif amount > balance:
            print("Insufficient Balance")
        else:
            balance -= amount
            print(f"Withdraw amount {amount}")
            print(f"New balance {balance}")

    elif choice == 4:
        print("\nThank you for using the ATM")
        print(f"Final balance {balance}")
        break

    else:
        print("Invalid choice! Please enter the valid choice.")
        
```

```python
# Q 8  Perfect Number & Divisors

num = int(input("Enter a number: "))

divisors = []
sum_of_divisors = 0

i = 1
while i < num: # 28
    if num % i == 0:
        divisors.append(i)
        sum_of_divisors += i
    i += 1

print("Proper Divisors of", num, ":", divisors)
print("Sum of Divisors:", sum_of_divisors)

if sum_of_divisors == num:
    print(f"{num} is a Perfect Number")
else:
    print(f"{num} is Not a Perfect Number")
```

```python
num = int(input("Enter the number: "))
temp = num
sum_var = 0

power_var = len(int(num))

while temp > 0:
    digit = temp % 10
    sum_var += digit ** power_var
    temp //= 10

if(sum_var == num):
    print(f"This is an armstorm number: {num}")
else:
    print(f"This is not an armstorm number: {num}")
    
```

```python
# HOLLO DIAMOND PATTERN

n = int(input("Enter number value: "))

## TOP
for i in range(1, n + 1):
    # 'j' to handle the spaces
    for j in range(1, n - i + 1):
        print(" ", end="")

    # 'k' handle to print '*' in the diamond border
    for k in range(1, 2 * i):
        if k == 1 or k == 2 * i - 1:
            print("*", end="")
        else:
            print(" ", end="")

    print()

## BOTTOM
for i in range(n - 1, 0, -1):
    # 'j' to handle the spaces
    for j in range(1, n - i + 1):
        print(" ", end="")

    # 'k' handle to print '*' in the diamond border
    for k in range(1, 2 * i):
        if k == 1 or k == 2 * i - 1:
            print("*", end="")
        else:
            print(" ", end="")

    print()
```

```python
n = 5

for i in range(n): 
    for j in range(n):
       
        x = min(i, j, n - 1 - i, n - 1 - j) # Finding the current layer
        
        start_val = n * n - (n - 2 * x) * (n - 2 * x) + 1 # Finding the first number of each layer
        L = n - 2 * x # Finding the layer's size
        
        if i == x: 
            # Top side (Left to Right)
            val = start_val + (j - x)
            
        elif j == n - 1 - x: 
            # Right side (Top to Bottom)
            val = start_val + (L - 1) + (i - x)
            
        elif i == n - 1 - x: 
            # Bottom side (Right to Left)
            val = start_val + 2 * (L - 1) + ((n - 1 - x) - j)
            
        else: 
            # Left side (Bottom to Top)
            val = start_val + 3 * (L - 1) + ((n - 1 - x) - i)
        print(f"{val:3}", end=" ")
        
    print()
```
