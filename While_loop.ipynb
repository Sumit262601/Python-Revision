```python
attempts = 3
correct_pin = "1206"

while attempts > 0:
    pin = input("Enter your four digits PIN: ")

    if pin.isdigit() and len(pin) == 4:
        if pin == correct_pin:
            print("Access granted.")
            break
    else:
        print("Please enter the four digits number!")
        
    attempts -= 1
    print(f"Wrong PIN. {attempts} tries left.")
else:
    print("Your account has been locked for to many requests.")
```

```python
## Fibonacci series
num = int(input("Enter the number: "))
a, b = 0, 1
count = 0

while count < num:
    print(a, end=" ")
    a, b = b, a + b
    # b = a, b = a+b
    count += 1

```

```python
# 1. Check the prime or not prime number
num = int(input("Enter the number: "))

if num <= 1:
    print(f"{num} is not valid")
else:
    i = 2
    is_prime = True

    while i * i <= num:
        if num % i == 0:
            is_prime = False
            break
        i += 1

    if is_prime:
        print(f"{num} is prime number")
    else:
        print(f"{num} is not a prime number")

    

```
